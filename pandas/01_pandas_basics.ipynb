 # 📘 Pandas Series

A **Pandas Series** is a **one-dimensional labeled array** that can store data of any type, such as integers, floats, strings, or even Python objects.

Think of a Series as a **single column in a spreadsheet** or a **dictionary with ordered values**, where each value has an associated **index**.

## Key Characteristics

- One-dimensional data structure.
- Each element has an **index (label)**.
- Can store homogeneous or heterogeneous data.
- Built on top of NumPy arrays.
- Supports fast indexing, slicing, and vectorized operations.

## Syntax

```python
pd.Series(data, index=None, dtype=None, name=None)
```

## Example

```python
import pandas as pd

marks = pd.Series([85, 90, 78, 92])

print(marks)
```

**Output**

```text
0    85
1    90
2    78
3    92
dtype: int64
```

## Common Uses

- Storing a single column of data.
- Time-series data.
- Statistical analysis.
- Data preprocessing.
- Building DataFrames.

> **Note:** A **DataFrame** is simply a collection of one or more **Series** sharing the same index.

````python
s = pd.Series([1, 3, 5, 7, 9])
s
````

````python
s = pd.Series([1 , 3, 5, 7, 9], index=['a', 'b', 'c', 'd', 'e'])
s
````

# 📘 Pandas DataFrame

A **Pandas DataFrame** is a **two-dimensional labeled data structure** used to store and manipulate tabular data. It consists of **rows** and **columns**, similar to an Excel spreadsheet or SQL table.

A DataFrame is built from one or more **Pandas Series**, where each column is a Series sharing the same index.

## Key Characteristics

- Two-dimensional (rows and columns).
- Each column can have a different data type.
- Columns have labels (column names).
- Rows have labels (index).
- Built on top of NumPy arrays.
- Supports fast indexing, filtering, sorting, grouping, and data analysis.

## Syntax

```python
pd.DataFrame(data, index=None, columns=None, dtype=None)
```

## Example

```python
import pandas as pd

students = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Age": [20, 22, 21],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(students)

print(df)
```

**Output**

```text
    Name  Age  Marks
0    Ali   20     85
1  Ahmed   22     90
2   Sara   21     88
```

## Common Uses

- Data cleaning and preprocessing.
- Reading and writing CSV, Excel, JSON, and SQL files.
- Data analysis and exploration.
- Filtering, sorting, and grouping data.
- Statistical analysis.
- Preparing datasets for Machine Learning.

## DataFrame vs Series

| Feature | Series | DataFrame |
|---------|--------|-----------|
| Dimension | 1D | 2D |
| Structure | Single column | Multiple columns |
| Index | ✅ | ✅ |
| Column Names | ❌ | ✅ |
| Can store multiple data types | ❌ (per Series) | ✅ |

> **Note:** A **Series** represents a single column of data, while a **DataFrame** is a collection of one or more Series organized into rows and columns.
