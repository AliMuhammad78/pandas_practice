```python
import pandas as pd

df = pd.read_csv("iris.csv")
```

```python
# isna()
# isnull()
# Returns True for missing (NaN) values.

df.isna()
```

```python
# notna()
# notnull()
# Returns True for non-missing values.

df.notna()
```

## dropna()



```python
df.dropna() # Removes rows containing missing values.
```

```python
df.dropna(axis=1) # remove columns with missing values 
```

## fillna()
 Replaces missing values with a specified value.

 > inplace = True , if you write this as argument, it will change teh original df 



```python
df.fillna(0)
```
