# 📌 Difference Between `loc[]` and `iloc[]`

| Feature | `loc[]` | `iloc[]` |
|---------|----------|-----------|
| Selection Type | Uses **labels (names)** | Uses **integer positions** |
| Row Selection | Row names/index labels | Row numbers (0, 1, 2, ...) |
| Column Selection | Column names | Column positions |
| Slice End | **Inclusive** | **Exclusive** |
| Best Used When | You know the row/column names | You know the row/column positions |

## Example

```python
import pandas as pd

df = pd.DataFrame(
    {
        "Name": ["Ali", "Sara", "Ahmed"],
        "Age": [20, 21, 22]
    },
    index=["A", "B", "C"]
)
```

### `loc[]` (Uses Labels)

```python
df.loc["A"]          # First row (label A)

df.loc[:, "Name"]    # Name column
```

### `iloc[]` (Uses Positions)

```python
df.iloc[0]           # First row (position 0)

df.iloc[:, 0]        # First column (position 0)
```

> **Easy Way to Remember**
>
> - **`loc` = Label-based indexing** (uses row/column **names**)
> - **`iloc` = Integer-location indexing** (uses row/column **positions**)

#### select first row 

````python
df.iloc[0]
````

#### last row 

````python
df.iloc[-1]
````

#### first 5 rows 

````python
df.iloc[:5]
````

#### 10 to 20 rows 

````python
df.iloc[10 : 20 ]
````

### Seelect first column

````python
df.iloc[:, 0]
````

### select specific value 

#

````python
df.iloc[5, 3] # row 5 , column 3
````

### Select rows and columns together 

````python
df.iloc[0:5, 0:3] # frist 5 rows and first 3 columns
````

### select specific rows 

````python
df.iloc[[0, 10, 50]]
````
