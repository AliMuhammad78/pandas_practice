```python
import pandas as pd

df = pd.read_csv("iris.csv")
```

# &

```python

# Returns rows where BOTH conditions are True.

df[(df["sepal_length"] > 5) & (df["petal_length"] > 4)]
```

# |

```python

# Returns rows where AT LEAST ONE condition is True.

df[(df["species"] == "Iris-setosa") | (df["species"] == "Iris-virginica")]
```

# ~ negation

```python

# Reverses (negates) a boolean condition.

df[~(df["species"] == "Iris-setosa")] 
# Returns all flowers except setosa.
```

---

```python
df[
    (df["species"] == "Iris-setosa") &
    (df["sepal_length"] > 5)
]
# Returns all setosa flowers with sepal_length > 5
```

---

# isin()
Checks if values exist in a given list.



```python
df[df["species"].isin(["Iris-setosa", "Iris-virginica"])]
# Returns only rows where species is setosa or virginica.
```

# between()
 Checks if values lie within a specified range.



```python
df[df["sepal_length"].between(5, 6)]
# where sepal_length is between 5 and 6 (inclusive).
```

# query()
 Filters rows using a query string.



```python
df.query("species == 'Iris-setosa'")
```

# converting from float to integer


```python
df['sepal_length'] = df['sepal_length'].astype('int')
df
```
