# groupby()
> Groups data based on one or more columns and performs operations on each group.

#### Concept:
 1. Split   -> Split the data into groups.
 2. Apply   -> Apply a function (mean, sum, count, etc.) to each group.
 3. Combine -> Combine the results into a new DataFrame/Series.

```python
import pandas as pd

df = pd.read_csv("iris.csv")

# Average sepal length for each species
print(df.groupby("species")["sepal_length"].mean())
```

# agg()
Applies one or more aggregate functions to grouped data.

```python
print(
    df.groupby("species")["sepal_length"].agg(["mean", "min", "max"])
)
```

# pd.concat()
Combines DataFrames by stacking rows or columns.

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Ali", "Sara"],
    "Age": [20, 21]
})

df2 = pd.DataFrame({
    "Name": ["Ahmed", "Ayesha"],
    "Age": [22, 23]
})

result = pd.concat([df1, df2])

print(result)
```

# pd.merge()
 Combines DataFrames using a common column (similar to SQL JOIN).

```python
import pandas as pd

students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [85, 90, 88]
})

result = pd.merge(students, marks, on="ID")

print(result)
```
