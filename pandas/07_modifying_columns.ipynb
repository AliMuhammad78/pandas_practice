## Add a New Column

```python
import pandas as pd

df = pd.read_csv("iris.csv")
```

```python
df["sepal_area"] = df['sepal_length'] * df['sepal_width']
df.head()
```

## Dropping rows and columns
##### drop()

```python

# Removes rows or columns.

# Remove a column
df.drop("species", axis=1)

# Remove the first row
df.drop(0, axis=0)
```

```python
df.head()
```

### remane column

```python
df.rename(columns={"species": "flower_species"},inplace=True)
```

### Sorting values 

```python
df.sort_values(by="sepal_length", ascending=False)
```
