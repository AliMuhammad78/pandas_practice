## `pd.pivot_table()`

```python
# pd.pivot_table()
# Creates a summary table by grouping data and applying an aggregate function.
#
# Similar to Excel Pivot Tables.
```

 

### Quick Summary

| Parameter | Description |
|-----------|-------------|
| `data` | DataFrame to summarize |
| `values` | Column to aggregate |
| `index` | Row labels (grouping column) |
| `columns` | Creates column groups (optional) |
| `aggfunc` | Aggregate function (`mean`, `sum`, `count`, `max`, etc.) |

> **Remember:** `groupby()` groups data and returns grouped objects, while `pivot_table()` directly creates a summarized table, similar to an Excel Pivot Table.

````python

import pandas as pd

df = pd.read_csv("iris.csv")

pivot = pd.pivot_table(
    df,
    values="sepal_length",    # Column to summarize
    index="species",          # Rows
    aggfunc="mean"            # Aggregation function
)

print(pivot)
````

### Using Multiple Aggregation Functions

````python
pivot = pd.pivot_table(
    df,
    values="petal_length",
    index="species",
    aggfunc=["mean", "max", "min"]
)

print(pivot)
````
